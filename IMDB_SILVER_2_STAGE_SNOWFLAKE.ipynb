```python
# Databricks DLT Pipeline: Silver to Snowflake STAGE

import dlt
from pyspark.sql.functions import *
from pyspark.sql.window import Window
from pyspark.sql.types import *


# ============================================
# CONFIGURATION
# ============================================

SNOWFLAKE_OPTIONS = {
    "sfUrl": "RCVOYBU-ZSC95331.snowflakecomputing.com",
    "sfUser": "IMDB_USER",
    "sfPassword": "IMDB",
    "sfDatabase": "IMDB_DB",
    "sfSchema": "STAGE",
    "sfWarehouse": "IMDB_WH",
    "sfRole": "IMDB_ROLE"
}

SOURCE_SCHEMA = "imdb_final_project.silver"

# ============================================
# SILVER TITLE BASICS
# ============================================

@dlt.table(name="silver_title_basics")
def silver_title_basics():
    df = spark.table(f"{SOURCE_SCHEMA}.silver_title_basics")
    
    df.write.format("snowflake").options(**SNOWFLAKE_OPTIONS) \
        .option("dbtable", "SILVER_TITLE_BASICS").mode("overwrite").save()
    
    return df

# ============================================
# SILVER TITLE RATINGS
# ============================================

@dlt.table(name="silver_title_ratings")
def silver_title_ratings():
    df = spark.table(f"{SOURCE_SCHEMA}.silver_title_ratings")
    
    df.write.format("snowflake").options(**SNOWFLAKE_OPTIONS) \
        .option("dbtable", "SILVER_TITLE_RATINGS").mode("overwrite").save()
    
    return df

# ============================================
# SILVER TITLE EPISODE
# ============================================

@dlt.table(name="silver_title_episode")
def silver_title_episode():
    df = spark.table(f"{SOURCE_SCHEMA}.silver_title_episode")
    
    df.write.format("snowflake").options(**SNOWFLAKE_OPTIONS) \
        .option("dbtable", "SILVER_TITLE_EPISODE").mode("overwrite").save()
    
    return df

# ============================================
# SILVER TITLE AKAS
# ============================================

@dlt.table(name="silver_title_akas")
def silver_title_akas():
    df = spark.table(f"{SOURCE_SCHEMA}.silver_title_akas")
    
    df.write.format("snowflake").options(**SNOWFLAKE_OPTIONS) \
        .option("dbtable", "SILVER_TITLE_AKAS").mode("overwrite").save()
    
    return df

# ============================================
# SILVER TITLE PRINCIPALS
# ============================================

@dlt.table(name="silver_title_principals")
def silver_title_principals():
    df = spark.table(f"{SOURCE_SCHEMA}.silver_title_principals")
    
    df.write.format("snowflake").options(**SNOWFLAKE_OPTIONS) \
        .option("dbtable", "SILVER_TITLE_PRINCIPALS").mode("overwrite").save()
    
    return df

# ============================================
# SILVER TITLE CREW
# ============================================

@dlt.table(name="silver_title_crew")
def silver_title_crew():
    df = spark.table(f"{SOURCE_SCHEMA}.silver_title_crew")
    
    df.write.format("snowflake").options(**SNOWFLAKE_OPTIONS) \
        .option("dbtable", "SILVER_TITLE_CREW").mode("overwrite").save()
    
    return df

# ============================================
# SILVER NAME BASICS
# ============================================

@dlt.table(name="silver_name_basics")
def silver_name_basics():
    df = spark.table(f"{SOURCE_SCHEMA}.silver_name_basics")
    
    df.write.format("snowflake").options(**SNOWFLAKE_OPTIONS) \
        .option("dbtable", "SILVER_NAME_BASICS").mode("overwrite").save()
    
    return df

# ============================================
# SILVER TITLE REGION
# ============================================

@dlt.table(name="silver_title_region")
def silver_title_region():
    df = spark.table(f"{SOURCE_SCHEMA}.silver_title_region")
    
    df.write.format("snowflake").options(**SNOWFLAKE_OPTIONS) \
        .option("dbtable", "SILVER_TITLE_REGION").mode("overwrite").save()
    
    return df

# ============================================
# SILVER TITLE LANGUAGE CODES
# ============================================

@dlt.table(name="silver_title_language_codes")
def silver_title_language_codes():
    df = spark.table(f"{SOURCE_SCHEMA}.silver_title_language_codes")
    
    df.write.format("snowflake").options(**SNOWFLAKE_OPTIONS) \
        .option("dbtable", "SILVER_TITLE_LANGUAGE_CODES").mode("overwrite").save()
    
    return df
```
